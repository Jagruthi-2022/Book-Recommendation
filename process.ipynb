```python
Ratings= pd.read_csv("Ratings.csv",on_bad_lines="skip",encoding="latin-1",low_memory=True)
Ratings.head()
```

```python
Ratings.shape
```

```python
Ratings.columns
```

```python
Ratings.rename(columns={
        "User-ID":"user_id",
        "Book-Rating":"rating"},
inplace=True)
```

```python
Ratings.head()
```

```python
print(books.shape,Ratings.shape)
```

```python
Ratings["user_id"].value_counts()
```

```python
Ratings["user_id"].value_counts().shape
```

```python
x=Ratings["user_id"].value_counts()>200
```

```python
x.shape
```

```python
x[x].shape
```

```python
y=x[x].index
```

```python
y
```

```python
ratings = Ratings[Ratings["user_id"].isin(y)]
```

```python
ratings.head()
```

```python
ratings.shape
```

```python
book_ratings= ratings.merge(books,on="ISBN")
```

```python
book_ratings.head()
```

```python
book_ratings.shape
```

```python
num_rating = book_ratings.groupby("title")["rating"].count().reset_index()
```

```python
num_rating.head()
```

```python
num_rating.rename(columns={
        "rating": "num_of_rating"},
inplace=True)
```

```python
num_rating.head()
```

```python
final_rating= book_ratings.merge(num_rating,on="title")
```

```python
final_rating.head()
```

```python
final_rating.shape
```

```python
final_rating= final_rating[final_rating["num_of_rating"]>=50]
```

```python
final_rating.head()
```

```python
final_rating.shape
```

```python
final_rating.drop_duplicates(["user_id","title"],inplace=True)
```

```python
final_rating.shape
```

```python
final_rating.head()
```

```python
book_pivot=final_rating.pivot_table(columns="user_id",index="title",values="rating")
```

```python
book_pivot
```

```python
book_pivot.shape
```

```python
book_pivot.fillna(0,inplace=True)
book_pivot
```

```python
from scipy.sparse import csr_matrix
```

```python
book_sparse = csr_matrix(book_pivot)
```

```python
book_sparse
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(algorithm="brute",metric="cosine")
```

```python
model.fit(book_sparse)
```

```python
distance,suggestion = model.kneighbors(book_pivot.iloc[237,:].values.reshape(1,-1),n_neighbors=6)
```

```python
distance
```

```python
suggestion
```

```python
for i in range(len(suggestion)):
        print(book_pivot.index[suggestion[i]])
```
